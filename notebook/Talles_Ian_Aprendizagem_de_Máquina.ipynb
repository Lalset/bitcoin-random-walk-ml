# Análise Histórica do Preço do Bitcoin com Aprendizagem de Máquina

Este projeto tem como objetivo analisar o comportamento histórico do preço do Bitcoin utilizando técnicas de análise de dados e visualização gráfica. O dataset utilizado contém informações históricas de negociação do Bitcoin ao longo do tempo.

## Carregamento do Dataset

O dataset utilizado foi obtido da plataforma Kaggle, contendo dados históricos do preço do Bitcoin. As informações incluem preço de abertura, fechamento, máximo, mínimo e volume de negociação.

Os dados foram carregados utilizando a biblioteca Pandas para permitir manipulação e análise eficiente.

```python
import kagglehub

path = kagglehub.dataset_download("mczielinski/bitcoin-historical-data")

print("Caminho", path)
```

```python
import os

os.listdir(path)
```

```python
import pandas as pd

file_path = os.path.join(path, "btcusd_1-min_data.csv")

df = pd.read_csv(file_path)

df.head()
```

```python
df.info()
```

```python
df.head()
```

```python
df.describe()
```

## Conversão de Timestamp

Os registros temporais estavam originalmente no formato Unix Timestamp. Dessa forma, foi realizada a conversão desses valores para o formato de data padrão, permitindo análises temporais adequadas.

```python
df["Date"] = pd.to_datetime(df["Timestamp"], unit="s")

df.head()
```

## Limpeza dos Dados

Nesta etapa, foram removidos registros contendo valores ausentes. Essa etapa é fundamental para garantir a qualidade e confiabilidade das análises realizadas posteriormente.

```python
df = df.dropna()
df.isnull().sum()
```

```python
print(df.columns.tolist())
```

## Agregação Temporal dos Dados

Os dados originais estavam registrados em intervalos de minutos, o que resultava em um grande volume de informações. Dessa forma, foi realizada uma agregação temporal para transformar os dados em registros diários.

Essa transformação reduz a complexidade computacional e facilita a visualização e análise do comportamento do preço ao longo do tempo.

```python
df_daily = df.resample("D", on="Date").agg({
    "Open": "first",
    "High": "max",
    "Low": "min",
    "Close": "last",
    "Volume": "sum"
}).dropna()

df_daily = df_daily.reset_index()

df_daily.head()
```

## Visualização do Preço do Bitcoin

Após o pré-processamento dos dados, foi gerado um gráfico representando a evolução histórica do preço de fechamento do Bitcoin.

A biblioteca Plotly foi utilizada para gerar visualizações interativas, permitindo melhor compreensão das tendências ao longo do tempo.

```python
import plotly.express as px

fig = px.line(
    df_daily,
    x="Date",
    y="Close",
    title="Preço diário de fechamento do Bitcoin"
)

fig.show()
```

## Dimensão do Dataset após Agregação

Após a agregação temporal dos dados, o dataset passou a conter 5230 registros diários e 10 atributos. Cada linha representa um dia de negociação do Bitcoin, contendo informações como abertura, máxima, mínima, fechamento e volume.

Essa transformação permitiu reduzir o volume original dos dados, que estavam em frequência de minutos, tornando a análise mais eficiente e adequada para visualizações e modelos de aprendizagem de máquina.

```python
df_daily.shape
```

## Cálculo dos Retornos Diários

Além do preço de fechamento, também foram calculados os retornos diários do Bitcoin. O retorno representa a variação percentual do preço entre dois dias consecutivos.

Também foi calculado o retorno logarítmico, frequentemente utilizado em análises de séries temporais financeiras por facilitar a interpretação das variações relativas ao longo do tempo.

```python
import numpy as np

df_daily["Return"] = df_daily["Close"].pct_change()
df_daily["Log_Return"] = np.log(df_daily["Close"] / df_daily["Close"].shift(1))

df_daily = df_daily.dropna()

df_daily.head()
```

## Visualização dos Retornos Logarítmicos

Após o cálculo dos retornos logarítmicos, foi realizada a visualização da variação diária do Bitcoin ao longo do tempo.

Esse tipo de visualização permite identificar períodos de alta volatilidade, nos quais ocorrem variações mais intensas no preço do ativo.

```python
import plotly.express as px

fig = px.line(
    df_daily,
    x="Date",
    y="Log_Return",
    title="Retorno Logarítmico Diário do Bitcoin"
)

fig.show()
```

## Distribuição dos Retornos

Foi gerado um histograma dos retornos logarítmicos com o objetivo de observar a distribuição das variações diárias do preço do Bitcoin.

Essa análise permite verificar se os retornos apresentam concentração próxima de zero ou se existem valores extremos associados a períodos de alta volatilidade.

```python
fig = px.histogram(
    df_daily,
    x="Log_Return",
    nbins=100,
    title="Distribuição dos Retornos Logarítmicos"
)

fig.show()
```

## Cálculo das Médias Móveis

As médias móveis são utilizadas para suavizar variações de curto prazo e destacar tendências ao longo do tempo.

Neste trabalho, foram calculadas médias móveis de 7 e 30 dias, representando tendências de curto e médio prazo.

```python
df_daily["MA_7"] = df_daily["Close"].rolling(window=7).mean()
df_daily["MA_30"] = df_daily["Close"].rolling(window=30).mean()

fig = px.line(
    df_daily,
    x="Date",
    y=["Close", "MA_7", "MA_30"],
    title="Preço do Bitcoin com Médias Móveis"
)

fig.show()
```

## Análise de Autocorrelação

A autocorrelação foi calculada com o objetivo de verificar se os retornos do Bitcoin apresentam dependência temporal.

Essa análise permite identificar se valores passados influenciam valores futuros, sendo um aspecto importante na verificação do comportamento de caminhada aleatória (random walk).

```python
import plotly.express as px
import pandas as pd

lag_max = 30

acf_values = [
    df_daily["Log_Return"].autocorr(lag=i)
    for i in range(lag_max)
]

acf_df = pd.DataFrame({
    "Lag": range(lag_max),
    "Autocorrelation": acf_values
})

fig = px.bar(
    acf_df,
    x="Lag",
    y="Autocorrelation",
    title="Autocorrelação dos Retornos Logarítmicos"
)

fig.show()
```

## Cálculo do Expoente de Hurst

O expoente de Hurst foi calculado com o objetivo de avaliar a presença de dependência de longo prazo na série temporal.

Esse indicador permite identificar se a série apresenta comportamento aleatório, tendência persistente ou reversão à média, sendo amplamente utilizado em análises financeiras e estudos de séries temporais.

```python
!pip install hurst
```

O cálculo inicial do Expoente de Hurst utilizando o preço de fechamento bruto resultou em valor acima do intervalo esperado. Por esse motivo, a análise foi ajustada para utilizar os retornos logarítmicos, que são mais adequados para séries financeiras por representarem variações relativas entre períodos consecutivos.

```python
from hurst import compute_Hc

serie = df_daily["Log_Return"].dropna().values

H, c, data = compute_Hc(
    serie,
    kind="change",
    simplified=True
)

print("Expoente de Hurst:", H)
```

## Cálculo do Expoente de Hurst

O expoente de Hurst foi calculado utilizando os retornos logarítmicos do preço de fechamento do Bitcoin, com o objetivo de avaliar a presença de dependência de longo prazo na série temporal.

O valor obtido para o expoente de Hurst foi aproximadamente H = 0.648. Esse resultado indica comportamento persistente, sugerindo que a série apresenta dependência temporal de longo prazo.

Esse comportamento indica que variações passadas podem influenciar tendências futuras, embora não de forma determinística, sendo característico de séries financeiras reais.

## Preparação dos Dados para Machine Learning

Para realizar a previsão do preço do Bitcoin, foram criadas variáveis baseadas em valores passados da série temporal. Essas variáveis, conhecidas como lags, representam valores anteriores que podem ser utilizados como entrada para modelos de aprendizagem de máquina.

Essa abordagem permite que o modelo identifique padrões temporais presentes nos dados históricos.

```python
df_ml = df_daily.copy()

df_ml["Close_lag1"] = df_ml["Close"].shift(1)
df_ml["Close_lag2"] = df_ml["Close"].shift(2)
df_ml["Close_lag3"] = df_ml["Close"].shift(3)

df_ml = df_ml.dropna()

df_ml.head()
```

```python
df_ml.shape
```

## Divisão entre Treino e Teste

Como se trata de uma série temporal, os dados foram divididos mantendo a ordem cronológica. Os primeiros 80% dos registros foram utilizados para treinamento do modelo, enquanto os 20% finais foram usados para teste.

Essa abordagem evita o embaralhamento dos dados e simula melhor um cenário real de previsão.

```python
X = df_ml[["Close_lag1", "Close_lag2", "Close_lag3"]]
y = df_ml["Close"]

split = int(len(df_ml) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

print("Treino:", X_train.shape)
print("Teste:", X_test.shape)
```

## Treinamento do Modelo Random Forest

Foi utilizado o algoritmo Random Forest Regressor para prever o preço de fechamento do Bitcoin. Esse modelo combina várias árvores de decisão para gerar uma previsão mais robusta e reduzir o risco de sobreajuste em relação a uma única árvore.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred) ** 0.5

print("MAE:", mae)
print("RMSE:", rmse)
```

```python
df_daily["Close"].mean()
```

## Ajuste da Variável Alvo

Inicialmente, o modelo foi treinado para prever diretamente o preço de fechamento do Bitcoin. No entanto, séries financeiras apresentam crescimento ao longo do tempo, o que pode dificultar a aprendizagem do modelo.

Dessa forma, foi adotada uma abordagem alternativa, utilizando os retornos logarítmicos como variável alvo, permitindo uma modelagem mais adequada das variações relativas da série temporal.

```python
df_ml2 = df_daily.copy()

df_ml2["Return_lag1"] = df_ml2["Log_Return"].shift(1)
df_ml2["Return_lag2"] = df_ml2["Log_Return"].shift(2)
df_ml2["Return_lag3"] = df_ml2["Log_Return"].shift(3)

df_ml2 = df_ml2.dropna()
```

```python
X = df_ml2[
    [
        "Return_lag1",
        "Return_lag2",
        "Return_lag3"
    ]
]

y = df_ml2["Log_Return"]
```

```python
split = int(len(df_ml2) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]
```

## Importância das Variáveis

Foi realizada a análise de importância das variáveis utilizadas no modelo Random Forest, com o objetivo de identificar quais valores passados exercem maior influência na previsão dos retornos futuros.

Essa análise permite compreender melhor o comportamento temporal da série.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred) ** 0.5

print("MAE:", mae)
print("RMSE:", rmse)
```

## Avaliação do Modelo com Retornos Logarítmicos

Após o ajuste da variável alvo para retornos logarítmicos, o modelo Random Forest apresentou desempenho significativamente melhor.

Os resultados obtidos foram:

- MAE ≈ 0.01987
- RMSE ≈ 0.02718

Esses valores indicam que o modelo consegue prever variações relativas com erro médio aproximado de 2%, demonstrando capacidade moderada de capturar padrões presentes na série temporal.

```python
import pandas as pd
import plotly.express as px

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

fig = px.bar(
    importance,
    x="Feature",
    y="Importance",
    title="Importância das Variáveis no Modelo"
)

fig.show()

importance
```

## Importância das Variáveis

A Figura X apresenta a importância relativa das variáveis utilizadas no modelo Random Forest.

Observa-se que o retorno do dia anterior (Return_lag1) apresentou maior relevância na previsão dos retornos futuros, seguido pelos retornos dos dois e três dias anteriores.

Esse comportamento indica a presença de dependência temporal de curto prazo, sugerindo que eventos recentes possuem maior influência na dinâmica da série temporal.

## Comparação entre Retornos Reais e Previstos

Foi realizada a comparação entre os retornos reais e os retornos previstos pelo modelo Random Forest.

Essa visualização permite avaliar a capacidade do modelo em acompanhar as variações da série temporal.

```python
results = pd.DataFrame({
    "Date": df_ml2["Date"].iloc[split:],
    "Real": y_test,
    "Predito": y_pred
})

fig = px.line(
    results,
    x="Date",
    y=["Real", "Predito"],
    title="Retornos Reais vs Retornos Previstos"
)

fig.show()
```

## Comparação entre Retornos Reais e Previstos

A Figura X apresenta a comparação entre os retornos reais e os retornos previstos pelo modelo Random Forest.

Observa-se que o modelo foi capaz de acompanhar o comportamento geral da série temporal, demonstrando capacidade de capturar padrões básicos presentes nos dados históricos.

No entanto, o modelo apresentou maior dificuldade na previsão de eventos extremos, especialmente em períodos de alta volatilidade. Esse comportamento é esperado em séries financeiras, nas quais variações abruptas são frequentemente associadas a fatores externos e imprevisíveis.

## Síntese dos Resultados Obtidos

Os resultados obtidos indicam que o preço do Bitcoin apresenta características típicas de séries financeiras, incluindo alta volatilidade e presença de dependência temporal de curto prazo.

A análise de autocorrelação indicou baixa dependência temporal imediata, enquanto o cálculo do expoente de Hurst revelou comportamento persistente de longo prazo (H ≈ 0.648).

O modelo Random Forest foi capaz de prever retornos logarítmicos com erro médio aproximado de 2%, demonstrando capacidade moderada de capturar padrões presentes na série temporal. Entretanto, o modelo apresentou maior dificuldade na previsão de eventos extremos.
